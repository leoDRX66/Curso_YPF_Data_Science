¿Porque ocurre esto?

Esto ocurre porque, aunque se usó "LabelEncoder" con "City_Category", esta herramienta de scikit-learn esta diseñada principalmente para codificar las variables **target** (las que vamos a predecir, usualmente en una columna), no los **features** (osea la variables predictivas para entrenar al entorno)

### Solucion

# Problema con get_dummies()
En este casos pd.get_dummies por defecto devuelve True/False (booleanos) en versiones recientes de Pandas y crea una columna por cada categoría. Esto genera "multicolinealidad" (osea dos columnas del dataset están aportando exactamente la misma informacion al modelo)  

```python
dummies_gender=pd.get_dummies(blackfriday["Gender"], drop_first=True, dtype=int)
display(dummies_gender.head())
```

```python
# Para unirlo al dataframe original (como se intentó hacer):
# Nota: get_dummies con drop_first=True sobre Gender (que tiene M, F y NaNs)
# devolverá solo una columna (por ejemplo 'M'). 
# El enfoque original creaba dos columnas ('female', 'male') sin drop_first.
# Mantendremos el enfoque original de agregar dos columnas, pero casteadas a int.
blackfriday[['famale','male']]=pd.get_dummies(blackfriday["Gender"], dtype=int)
blackfriday.head()
```

```python
from sklearn.preprocessing import OneHotEncoder

# Vemos los valores únicos: array(['F', nan, 'M'], dtype=object)
print(blackfriday['Gender'].unique())

# ANOTACIÓN CORREGIDA: si hago drop first despues no puede pedir categorias
# El problema: Si usamos drop='first' en OneHotEncoder, elimina una categoría para evitar colinealidad.
# Si luego hacemos `encoder.categories_`, nos mostrará TODAS las categorías originales que aprendió,
# y al intentar mapear los nombres de las columnas (con len=3) a la matriz transformada (con len=2), fallará.
# La solución: Usar el método oficial `get_feature_names_out()` en lugar de intentar extraer a mano `.categories_[0].tolist()`.

# RESOLUCIÓN CORRECTA:
# 1. Tratamos los valores NaN antes (o usamos handle_unknown)
# Para este ejemplo, llenaremos los nulos con 'Unknown' momentáneamente o usamos un imputador.
# Vamos a seguir la lógica original pero de forma robusta.

gender_encoder_robust = OneHotEncoder(sparse_output=False, drop='first') # sparse_output en versiones nuevas de sklearn

# Ajustamos y transformamos (OJO: Gender tiene NaNs, sklearn puede quejarse si no los manejamos, 
# pero asumiremos que el df está limpio o llenamos los nulos con 'Desconocido')
blackfriday['Gender_Filled'] = blackfriday['Gender'].fillna('Unknown')
matriz_one_hot = gender_encoder_robust.fit_transform(blackfriday[['Gender_Filled']])

# Obtenemos los nombres de las columnas correctamente generados
niveles_correctos = gender_encoder_robust.get_feature_names_out(['Gender_Filled'])

# Creamos el DataFrame
one_hot_gender_df = pd.DataFrame(matriz_one_hot, columns=niveles_correctos)

# Unimos con el dataset original
new_df = pd.concat([blackfriday, one_hot_gender_df], axis=1)
display(new_df.head())
```

**`binning`**

```python
print(blackfriday.Age.min())
print(blackfriday.Age.max())
```

```python
# Nos asguramos de que sea numérica (el código original lo hace bien)
blackfriday["Age"] = pd.to_numeric(blackfriday["Age"], errors='coerce')
```

```python
# Definimos los cortes (bins) y las etiquetas (labels)
bin_age = [10, 17, 70, 80]
labels = ["Adolescente", "Adulto", "Anciano"]

# Aplicamos la categorización
age_categories = pd.cut(blackfriday["Age"], bins=bin_age, labels=labels)
display(age_categories.head())
```
