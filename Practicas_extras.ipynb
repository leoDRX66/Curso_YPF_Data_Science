````python
# Ejercicio 11
def promedio_produccion(lista_produccion):
    return sum(lista_produccion) / len(lista_produccion)
print(promedio_produccion([150.5, 160.2, 148.9, 155.0]))
````

````python
# Ejecicio 14
def calcular_energia_por_panel(energia_total, cantidad_paneles):
    try:
        return energia_total/cantidad_paneles
    except ZeroDivisionError:
        return 0

print("Con 10 paneles", calcular_energia_por_panel(1000, 10))
print("Con 0 paneles (Error manejado)", calcular_energia_por_panel(1000, 0))
````

````python
# Ejercicio 15
def evaluar_estacion(datos):
    promedio=sum(datos["presion"]) / len(datos["presion"])
    if promedio < 1000:
        return f"ALERTA: presion baja{datos['nombre']}. Promedio: {promedio}"
    return f"Estado normal en {datos['nombre']}."

datos_estacion={"nombre": "Estacion vineto sur", "presion": [955,990,992,998,991]}
print(evaluar_estacion(datos_estacion))
````
